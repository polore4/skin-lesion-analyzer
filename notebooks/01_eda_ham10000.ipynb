```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Configuración visual
sns.set_theme(style="whitegrid")

# Cargar los metadatos
metadata_path = '../data/HAM10000_metadata.csv'

if os.path.exists(metadata_path):
    df = pd.read_csv(metadata_path)
    print(f"Dataset cargado correctamente. Total de imágenes: {len(df)}")
    
    # Mostrar distribución de clases
    plt.figure(figsize=(10, 6))
    ax = sns.countplot(data=df, x='dx', order=df['dx'].value_counts().index, palette='viridis')
    plt.title('Distribución de Clases de Lesiones Cutáneas (HAM10000)', fontsize=14)
    plt.xlabel('Tipo de Lesión (Clase)', fontsize=12)
    plt.ylabel('Cantidad de Muestras', fontsize=12)
    
    # Añadir los números encima de las barras
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha='center', va='baseline', fontsize=10, color='black', xytext=(0, 5), 
                    textcoords='offset points')
    plt.show()
else:
    print(f"Error: No se encuentra el archivo en {metadata_path}. Asegúrate de colocar el CSV en la carpeta 'data/'.")
```
